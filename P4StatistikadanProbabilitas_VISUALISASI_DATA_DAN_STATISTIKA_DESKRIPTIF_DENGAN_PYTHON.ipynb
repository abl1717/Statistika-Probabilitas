```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Tabel Data
df = r"/dataTambahan/abil.xlsx"
dataraw = pd.read_excel(df)
display(
    dataraw.style
    .set_caption("📊 TABEL DATA NILAI MAHASISWA")
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('caption-side', 'top'),
                ('font-size', '18px'),
                ('font-weight', 'bold'),
                ('color', 'white'),
                ('background-color', '#03A6A1'),
                ('padding', '12px'),
                ('border-radius', '8px 8px 0 0')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('background-color', '#03A6A1'),
                ('color', 'white'),
                ('font-weight', 'bold'),
                ('text-align', 'center'),
                ('padding', '9px')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('text-align', 'center'),
                ('padding', '8px')
            ]
        },
        {
            'selector': 'tbody tr:nth-child(even)',
            'props': [
                ('background-color', '#FFE3BB')
            ]
        },
        {
            'selector': 'tbody tr:hover',
            'props': [
                ('background-color', '#FFA673')
            ]
        }
    ])
    .set_properties(**{
        'border': '1px solid #FFE3BB'
    })
)
```

```python
# Tabel Frekuensi

datafrq = pd.crosstab(
    index=dataraw['Grade'],
    columns='Frekuensi'
)

datafrq['Persentase'] = (
    datafrq['Frekuensi'] /
    datafrq['Frekuensi'].sum() * 100
).round(2)

display(
    datafrq.style
    .set_caption("📈 TABEL DISTRIBUSI FREKUENSI GRADE")
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('caption-side', 'top'),
                ('font-size', '18px'),
                ('font-weight', 'bold'),
                ('color', 'white'),
                ('background-color', '#FF4F0F'),
                ('padding', '12px'),
                ('border-radius', '8px 8px 0 0')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('background-color', '#FFA673'),
                ('color', '#FF4F0F'),
                ('font-weight', 'bold'),
                ('text-align', 'center'),
                ('padding', '9px')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('text-align', 'center'),
                ('padding', '8px')
            ]
        },
        {
            'selector': 'tbody tr:nth-child(even)',
            'props': [
                ('background-color', '#FFE3BB')
            ]
        }
    ])
    .set_properties(**{
        'border': '1px solid #FFA673'
    })
    .format({
        'Frekuensi': '{:.0f}',
        'Persentase': '{:.2f}%'
    })
)
```

```python
# Grafik Garis
plt.figure(figsize=(11, 6))

plt.plot(
    datafrq.index,
    datafrq['Frekuensi'],
    color='#03A6A1',
    marker='o',
    markerfacecolor='#FFE3BB',
    markeredgecolor='#FF4F0F',
    markeredgewidth=2.5,
    linewidth=3,
    markersize=10
)

for x, y in zip(datafrq.index, datafrq['Frekuensi']):
    plt.annotate(
        f'{int(y)}',
        (x, y),
        textcoords="offset points",
        xytext=(0, 12),
        ha='center',
        fontsize=11,
        fontweight='bold',
        color='#FF4F0F'
    )

plt.title(
    '📈 Distribusi Frekuensi Grade',
    fontsize=18,
    fontweight='bold',
    color='#03A6A1',
    pad=20
)

plt.xlabel(
    'Grade',
    fontsize=13,
    fontweight='bold',
    color='#FF4F0F'
)

plt.ylabel(
    'Frekuensi',
    fontsize=13,
    fontweight='bold',
    color='#03A6A1'
)

ax = plt.gca()
ax.set_facecolor('#FFE3BB')

plt.grid(
    True,
    linestyle='--',
    linewidth=0.8,
    color='#FFA673',
    alpha=0.6
)

plt.xticks(
    fontsize=11,
    fontweight='bold',
    color='#FF4F0F'
)

plt.yticks(
    fontsize=10,
    color='#03A6A1'
)

plt.show()
```

```python
# Grafik Batang
plt.figure(figsize=(11, 6))

bars = plt.bar(
    datafrq.index,
    datafrq['Frekuensi'],
    color='#03A6A1',
    edgecolor='#FF4F0F',
    linewidth=1.8,
    width=0.65
)

max_index = datafrq['Frekuensi'].idxmax()

for bar, grade in zip(bars, datafrq.index):

    if grade == max_index:
        bar.set_color('#FF4F0F')
        bar.set_edgecolor('#FF4F0F')

    tinggi = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        tinggi + 0.2,
        f'{int(tinggi)}',
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold',
        color='#FF4F0F'
    )

plt.title(
    '📊 Distribusi Frekuensi Grade Mahasiswa',
    fontsize=18,
    fontweight='bold',
    color='#03A6A1',
    pad=20
)

plt.xlabel(
    'Grade',
    fontsize=13,
    fontweight='bold',
    color='#FF4F0F'
)

plt.ylabel(
    'Jumlah Mahasiswa',
    fontsize=13,
    fontweight='bold',
    color='#03A6A1'
)

ax = plt.gca()
ax.set_facecolor('#FFE3BB')

plt.grid(
    axis='y',
    linestyle='--',
    linewidth=0.8,
    color='#FFA673',
    alpha=0.6
)

plt.xticks(
    fontsize=11,
    fontweight='bold',
    color='#FF4F0F'
)

plt.yticks(
    fontsize=10,
    color='#03A6A1'
)

plt.show()
```

```python
# Pie Chart
plt.figure(figsize=(9, 9))

warna_pie = [
    '#03A6A1',
    '#FFE3BB',
    '#FFA673',
    '#FF4F0F'
]

wedges, texts, autotexts = plt.pie(
    datafrq['Frekuensi'],
    labels=datafrq.index,
    colors=warna_pie[:len(datafrq)],
    autopct='%1.1f%%',
    startangle=90,
    counterclock=False,
    pctdistance=0.75,
    wedgeprops={
        'edgecolor': 'white',
        'linewidth': 3
    },
    textprops={
        'fontsize': 12,
        'fontweight': 'bold',
        'color': '#FF4F0F'
    }
)

plt.setp(
    autotexts,
    fontsize=11,
    fontweight='bold',
    color='#FF4F0F'
)

plt.title(
    '🥧 Persentase Distribusi Grade',
    fontsize=18,
    fontweight='bold',
    color='#03A6A1',
    pad=20
)

plt.show()
```

```python
# Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(
    dataraw["Final Score"],
    errors='coerce'
)

dt = dataraw["Final Score"]

stats = dt.describe()

stats["Standard Error"] = dt.sem()
stats["Median"] = dt.median()
stats["Mode"] = dt.mode().iloc[0]
stats["Variance"] = dt.var()
stats["Range"] = dt.max() - dt.min()
stats["Skewness"] = dt.skew()
stats["Kurtosis"] = dt.kurtosis()

stats_table = stats.to_frame(name="Nilai")

display(
    stats_table.style
    .set_caption("📋 STATISTIKA DESKRIPTIF FINAL SCORE")
    .set_table_styles([
        {
            'selector': 'caption',
            'props': [
                ('caption-side', 'top'),
                ('font-size', '18px'),
                ('font-weight', 'bold'),
                ('color', 'white'),
                ('background-color', '#03A6A1'),
                ('padding', '12px'),
                ('border-radius', '8px 8px 0 0')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('background-color', '#FFA673'),
                ('color', '#FF4F0F'),
                ('font-weight', 'bold'),
                ('text-align', 'center'),
                ('padding', '9px')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('text-align', 'center'),
                ('padding', '8px')
            ]
        },
        {
            'selector': 'tbody tr:nth-child(even)',
            'props': [
                ('background-color', '#FFE3BB')
            ]
        }
    ])
    .set_properties(**{
        'border': '1px solid #FFA673'
    })
    .format('{:.2f}')
)
```
